```python
import seaborn as sns              
import matplotlib.pyplot as plt    
import pandas as pd               
```

```python
penguins = pd.read_csv("penguins.csv")
penguins.head()
```

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.show()
#shows relationship between 2 numeric features
```

```python
sns.boxplot(data=penguins, x="species", y="flipper_length_mm")
plt.show()
#highlights distribution and outliners
```

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g", hue="sex", split=True)
plt.show()
#combines box plot with kde curve
```

```python
sns.pairplot(penguins, hue="species")
plt.show()
#shows pairwise relationship 
```

```python
sns.heatmap(penguins.corr(), annot=True, cmap="coolwarm")
plt.show()
#for visualising correlations
```

```python
sns.countplot(data=penguins, x="species", hue="sex")
plt.show()
#shows number of occurrences
```
