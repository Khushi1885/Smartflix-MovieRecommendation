```python
import qrcode
import tkinter as tk
from tkinter import ttk
from tkinter import filedialog
from PIL import Image, ImageTk

class QRCodeGeneratorGUI:
    def __init__(self, master):
        self.master = master
        master.title("QR Code Generator")

        self.data_label = ttk.Label(master, text="Data:")
        self.data_label.grid(row=0, column=0, padx=5, pady=5, sticky=tk.W)

        self.data_entry = ttk.Entry(master, width=40)
        self.data_entry.grid(row=0, column=1, padx=5, pady=5, sticky=tk.E)

        self.filename_label = ttk.Label(master, text="Filename:")
        self.filename_label.grid(row=1, column=0, padx=5, pady=5, sticky=tk.W)

        self.filename_entry = ttk.Entry(master, width=30)
        self.filename_entry.grid(row=1, column=1, padx=5, pady=5, sticky=tk.W)

        self.browse_button = ttk.Button(master, text="Browse", command=self.browse_file)
        self.browse_button.grid(row=1, column=2, padx=5, pady=5, sticky=tk.E)

        self.generate_button = ttk.Button(master, text="Generate QR Code", command=self.generate_qr)
        self.generate_button.grid(row=2, column=1, padx=5, pady=10, sticky=tk.E)

        self.image_label = ttk.Label(master)
        self.image_label.grid(row=3, column=0, columnspan=3, padx=5, pady=5)

        self.status_label = ttk.Label(master, text="")
        self.status_label.grid(row=4, column=0, columnspan=3, padx=5, pady=5)

    def browse_file(self):
        filename = filedialog.asksaveasfilename(defaultextension=".png")
        self.filename_entry.delete(0, tk.END)
        self.filename_entry.insert(0, filename)

    def generate_qr(self):
        data = self.data_entry.get()
        filename = self.filename_entry.get()

        if not data:
            self.status_label.config(text="Please enter data to encode.")
            return

        if not filename:
            self.status_label.config(text="Please enter a filename.")
            return

        try:
            qr = qrcode.QRCode(
                version=1,
                error_correction=qrcode.constants.ERROR_CORRECT_L,
                box_size=10,
                border=4,
            )
            qr.add_data(data)
            qr.make(fit=True)

            img = qr.make_image(fill_color="black", back_color="white")
            img.save(filename)

            self.status_label.config(text=f"QR code saved as {filename}")

            # Display the generated QR code in the GUI
            img = Image.open(filename)
            img = ImageTk.PhotoImage(img)
            self.image_label.config(image=img)
            self.image_label.image = img  # Keep a reference!

        except Exception as e:
            self.status_label.config(text=f"Error: {e}")


root = tk.Tk()
gui = QRCodeGeneratorGUI(root)
root.mainloop()

```
